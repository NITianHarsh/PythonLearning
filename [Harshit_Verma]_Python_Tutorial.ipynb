12. Horse Racing Puzzle
Problem: 25 horses and no stopwatch. use only 5 horses per race and find fastest top 3 horses.
13. so, for solving this the minimum races require is 7.
14. we can make 5 races by dividing horses in a group of 5 and get the winners of all those groups.
15. now, make a race of those 5 winner horses, to know the fastest and horse.
16. now, one more race is required to know the top 3 horses.

```python
import random

# create horses with hidden speeds
horses = list(range(1, 26))
random.shuffle(horses)
print("horses are: ", horses)

groups = [horses[i:i+5] for i in range(0, 25, 5)]
print(f'groups are {groups}')
#sort each group
for g in groups:
  g.sort(reverse=True)

winners = [g[0] for g in groups]
winners.sort(reverse=True)
print(f'winners are {winners}')

# reorder groups by winner speed
groups_sorted = sorted(groups, key=lambda x: x[0], reverse=True)

A = groups_sorted[0]
B = groups_sorted[1]
C = groups_sorted[2]

finalists = [A[1], A[2], B[0], B[1], C[0]]
finalists.sort(reverse=True)
print(f'finalists are {finalists}')

fastest = groups_sorted[0][0]  # A1
second = finalists[0]
third = finalists[1]

print("Top 3 horses:", fastest, second, third)
```

```python
def time_table():
  x = int(input("Please enter a number: "))
  for row in range(x):
    for col in range(x):
      print(f"{row * col:2}", end=" ")
    print()

time_table();
```
