## <u>Problem 1</u>

A list contains the average daily temperature(in degree Celsius) of a city over a particular week. Write a Python code to swap the highest and the lowest temperatures

```python
# A list containing average daily temperature over a week
temperatures = [34, 40, 29, 33, 42, 37, 39 ]
temperatures

# The expected output
# output_temperatures = [34, 40, 42, 33, 29, 37, 39]
```

```python
# Store the highest temperature
max_temp = max(temperatures)
print(max_temp)

# Index of the element with the highest temperature
max_temp_index = temperatures.index(max(temperatures))
print(max_temp_index)
```

```python
# Store the lowest temperature
min_temp = min(temperatures)
print(min_temp)

# Index of the element with the lowest temperature
min_temp_index = temperatures.index(min(temperatures))
print(min_temp_index)
```

```python
# Swap the highest and the lowest temperatures
temperatures[max_temp_index],temperatures[min_temp_index]=temperatures[min_temp_index],temperatures[max_temp_index]
```

### Measures of Averages

* Mean of a group is defined as the sum of the elements in the group divided by the number of elements in the group.
* Median is the middle number in a <b>sorted</b>, ascending or descending, list of numbers. If the list has $N$ numbers where $N$ is odd, then median is the element in the middle i.e, $(\frac{N+1}{2})^{th}$ element. If $N$ is even, then median is the mean of $(\frac{N}{2})^{th}$ and $(\frac{N}{2}+1)^{th}$ elements

```python
# List of product prices
prod_price_list = [400, 250, 800, 550, 600, 820, 720, 15000, 360,250,100]
```

```python
# Calculate the mean of the product prices. Use list functions. Also print the mean_price
mean_price = sum(prod_price_list)/len(prod_price_list)
print(mean_price)
```

## <u> Problem 3 </u>

Having a nested list sometimes might be a bit problematic. An individual was asked to collect the names of companies in the technology sector. While creating the list, by mistake the last three companies were subsumed in a list as shown below. You are required to get rid of the nesting

```python
# The list of tech companies curated by the individual
tech_companies = ['Qualcomm','Google','Apple',['Nvidia','Cisco','Samsung']]
print(tech_companies)

# The final list that we want
# correct_tech_companies = ['Qualcomm','Google','Apple','Nvidia','Cisco','Samsung']
```

```python
flat_list=[]
def flatten_list(tech_companies):
  for elements in tech_companies:
    if type(elements) == list:
      flatten_list(elements)
    else:
      flat_list.append(elements)

flatten_list(tech_companies)
print(flat_list)
```

```python
# Write your code below
# function
flat_list=[]
def flatten_list(tech_companies):
    # iterating over the data
    for element in tech_companies:
        # checking for list
        if type(element) == list:
            # calling the same function with current element as new argument
            flatten_list(element)
        else:
            flat_list.append(element)

# flattening the given list
flatten_list(tech_companies)

# printing the flat_list
print(flat_list)
```
