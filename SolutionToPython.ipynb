## Functions & Methods

### Problem 1

```python
def get_excel_column_number(col):
  string_len = len(col) - 1;
  last = ord(col[string_len]) - ord('A') + 1
  ans = last;

  while string_len > 0:
    last = ord(col[string_len-1]) - ord('A') + 1
    ans += last*(26**string_len)
    string_len -= 1;

  return ans;
```

```python
print(get_excel_column_number('BQ'))
print(get_excel_column_number('ALQ'))
```

### Problem 2

```python
def calculate_accuracy(actual, predicted):
    correct = 0

    for a, p in zip(actual, predicted):
        if a == p:
            correct += 1

    accuracy = (correct / len(actual)) * 100
    return accuracy
```

```python
actual_labels = ['Cat','Dog','Cat','Cat','Dog','Cat','Dog','Cat','Cat','Cat',
                 'Dog','Dog','Cat','Cat','Cat','Dog','Dog','Dog','Cat','Dog']

predicted_labels = ['Cat','Dog','Cat','Cat','Dog','Cat','Dog','Cat','Dog','Dog',
                    'Dog','Dog','Cat','Cat','Cat','Dog','Dog','Dog','Cat','Cat']

print("Accuracy:", calculate_accuracy(actual_labels, predicted_labels), "%")
```

### Problem 3

```python
def precision_recall(actual, predicted, label):
    true_positive = 0
    predicted_positive = 0
    actual_positive = 0

    for a, p in zip(actual, predicted):
        if p == label:
            predicted_positive += 1
        if a == label:
            actual_positive += 1
        if a == label and p == label:
            true_positive += 1

    precision = (true_positive / predicted_positive) * 100 if predicted_positive else 0
    recall = (true_positive / actual_positive) * 100 if actual_positive else 0

    return precision, recall
```

```python
p_cat, r_cat = precision_recall(actual_labels, predicted_labels, 'Cat')
print(f"Cat Precision: {p_cat:.2f}%")
print(f"Cat Recall: {r_cat:.2f}%")
```

```python
p_dog, r_dog = precision_recall(actual_labels, predicted_labels, 'Dog')
print(f"Dog Precision: {p_dog:.2f}%")
print(f"Dog Recall: {r_dog:.2f}%")
```

```python
def sentence_len_counts(sentence, length):
    return sum(1 for word in sentence.split() if len(word) == length)
```

### Problem 2

```python
prod_price_list = [400, 250, 800, 550, 600, 820, 720, 15000, 360,250,100]
```

```python
def calculate_median(lis):
    lis = sorted(lis)
    n = len(lis)

    if n % 2 == 0:
        return (lis[n//2] + lis[n//2 - 1]) / 2
    else:
        return lis[n//2]

def calculate_mean(lis):
  return sum(lis)//len(lis)
```

```python
print(calculate_median(prod_price_list))
print(calculate_mean(prod_price_list))
```
