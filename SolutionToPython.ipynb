### Problem 3

```python
apples = 45
oranges = 65
bananas = 30
```

```python
total = apples + bananas + oranges
```

```python
percent_apple = apples*100/total
percent_banana = bananas*100/total
percent_oranges = oranges*100/total
```

```python
print(f'apple percent: {round(percent_apple, 2)}')
print(f'banana percent: {round(percent_banana, 2)}')
print(f'orange percent: {round(percent_oranges, 2)}')
```
