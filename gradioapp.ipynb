# GRADIO

## 🔹 1. What is Gradio?

**Gradio is a Python library that lets you build web-based UIs for your models or functions with minimal code.**

### Why companies use it

* Rapid AI demos
* Internal ML tools
* Model testing dashboards
* Chatbot interfaces
* Proof of concepts (POCs)

Think of it as:
**Python function → Gradio → Instant web app**

---

## 🔹 2. Installation

```bash
pip install gradio
```

# 3. Core Mental Model

Gradio connects three things:

```
Input UI → Python Function → Output UI
```

This is the **heart of Gradio**.

---

# 4. Your First Gradio App

##  Basic Example

```python
import gradio as gr

def greet(name):
    return f"Hello {name}!"

app = gr.Interface(
    fn=greet,
    inputs="text",
    outputs="text"
)

app.launch()
```

### 🔍 What each part means

| Part       | Meaning              |
| ---------- | -------------------- |
| `fn`       | Your Python function |
| `inputs`   | Input component      |
| `outputs`  | Output component     |
| `launch()` | Starts web server    |

# 7. Interface vs Blocks

## 🔹 Interface

Use when:

* Simple apps
* One function
* Fast demo

✅ Less code
❌ Less control

---

## 🔹 Blocks (Professional Way)

Use when:

* Complex layout
* Multiple components
* Chat apps
* Production UI

✅ Full control
✅ Industry preferred

---

# 8. Blocks — Professional Pattern

## Example

```python
import gradio as gr

def greet(name):
    return f"Hello {name}"

with gr.Blocks() as demo:
    name = gr.Textbox(label="Name")
    output = gr.Textbox()

    btn = gr.Button("Submit")
    btn.click(fn=greet, inputs=name, outputs=output)

demo.launch()
```

---

## Why Blocks is Powerful

You can create:

* Rows
* Columns
* Tabs
* Accordions
* Chat interfaces
* Multi-step workflows

# 10. Chatbot UI (VERY HOT IN INDUSTRY)

## Basic Chatbot

```python
import gradio as gr

def chat(message, history):
    return "Echo: " + message

with gr.Blocks() as demo:
    chatbot = gr.Chatbot()
    msg = gr.Textbox()

    msg.submit(chat, [msg, chatbot], chatbot)

demo.launch()
```

Used in:

* ChatGPT clones
* Support bots
* AI assistants
