## Tutorial 2 - 23rd Feb. 2026

### 1. Boolean Variable

* Created using the `bool` type.
* Represents logical values: `True` or `False`.

```python
flag = bool(1)   # True
```

---

### 2. String Immutability

* Strings in Python are **immutable**.
* Any modification returns a new string object.

---

### 3. Commonly Used String Functions

#### 3.1 Case Conversion

* `lower()` — converts to lowercase
* `upper()` — converts to uppercase
* `title()` — capitalizes each word
* `capitalize()` — capitalizes first character

```python
text = "hello world"
text.lower()
text.upper()
text.title()
text.capitalize()
```

---

#### 3.2 Whitespace Handling

* `strip()` — removes leading and trailing whitespace
* `lstrip()` — removes leading whitespace
* `rstrip()` — removes trailing whitespace

```python
text = "  hello  "
text.strip()
```

---

#### 3.3 Replacement

* `replace(old, new)` — replaces substring occurrences

```python
text = "I like Java"
text.replace("Java", "Python")  # "I like Python"
```

---

#### 3.4 Splitting Strings

* `split(sep=None)` — splits string into a list

```python
text = "apple banana mango"
text.split()
```

**Output**

```python
['apple', 'banana', 'mango']
```

**With delimiter**

```python
data = "a,b,c"
data.split(",")
```

---

#### 3.5 Joining Strings

* `"separator".join(iterable)` — joins iterable into a string

```python
words = ['hello', 'world']
" ".join(words)   # "hello world"
```

**Key Point**

* `split()` → string → list
* `join()` → list → string

---

#### 3.6 Searching

* `find(sub)` — returns index or `-1` if not found
* `index(sub)` — returns index or raises `ValueError`

```python
text = "hello world"
text.find("world")   # 6

text = "hello"
text.index("e")      # 1
```

---

#### 3.7 Prefix and Suffix Checks

* `startswith(prefix)`
* `endswith(suffix)`

```python
"text".startswith("te")
"file.pdf".endswith(".pdf")
```

---

#### 3.8 Counting

* `count(sub)` — counts occurrences

```python
"banana".count("a")  # 3
```

---

#### 3.9 Validation Methods

* `isalpha()` — letters only
* `isdigit()` — digits only
* `isalnum()` — letters and digits
* `isspace()` — whitespace only

```python
"Hello".isalpha()
"123".isdigit()
"abc123".isalnum()
"   ".isspace()
```

### Python List
A **list** is an ordered, mutable collection that allows duplicate and heterogeneous elements.

```python
nums = [1, 2, 3]
mixed = [1, "hello", 3.5]
```

**Key Properties**

* Ordered
* Mutable
* Allows duplicates
* Dynamic size
* Supports indexing and slicing

---
### Indexing

```python
nums = [10, 20, 30]
nums[0]      # 10
nums[-1]     # 30
```

### Slicing

```python
nums = [0, 1, 2, 3, 4]
nums[1:4]    # [1, 2, 3]
```

---
### 1. `append(x)`

Adds element at the end.

```python
lst = [1, 2]
lst.append(3)   # [1, 2, 3]
```

---

### 2. `extend(iterable)`

Adds multiple elements.

```python
lst = [1, 2]
lst.extend([3, 4])   # [1, 2, 3, 4]
```

**Difference**

* `append()` → adds single item
* `extend()` → adds multiple items

---

### 3. `insert(index, x)`

Inserts element at specific position.

```python
lst = [1, 3]
lst.insert(1, 2)   # [1, 2, 3]
```

---

### 4. `remove(x)`

Removes first occurrence of value.

```python
lst = [1, 2, 2, 3]
lst.remove(2)   # [1, 2, 3]
```

⚠️ Raises error if value not present.

---

### 5. `pop([index])`

Removes and returns element.

```python
lst = [10, 20, 30]
lst.pop()      # removes 30
lst.pop(0)     # removes first element
```

---

### 6. `clear()`

Removes all elements.

```python
lst = [1, 2, 3]
lst.clear()   # []
```

---

### 7. `index(x)`

Returns index of first occurrence.

```python
lst = [10, 20, 30]
lst.index(20)   # 1
```

⚠️ Raises error if not found.

---

### 8. `count(x)`

Counts occurrences.

```python
lst = [1, 2, 2, 3]
lst.count(2)   # 2
```

---

### 9. `sort()`

Sorts list in-place.

```python
lst = [3, 1, 2]
lst.sort()   # [1, 2, 3]
```

**Descending**

```python
lst.sort(reverse=True)
```

**With key**

```python
words = ["apple", "kiwi", "banana"]
words.sort(key=len)
```

---

### 10. `reverse()`

Reverses list in-place.

```python
lst = [1, 2, 3]
lst.reverse()   # [3, 2, 1]
```

---

### 11. `copy()`

Creates shallow copy.

```python
lst = [1, 2]
new_lst = lst.copy()
```

---


### List Comprehension (Very Important)

Concise way to create lists.

```python
squares = [x*x for x in range(5)]
```

With condition:

```python
evens = [x for x in range(10) if x % 2 == 0]
```
